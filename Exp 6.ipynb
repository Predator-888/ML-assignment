```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

# Data Initialization

```python
d1 = {"Dhurandhar": [5, 4, 1, 5, 0],
      "Jailer-2": [1, 0, 5, 1, 4],
      "Drishyum": [4, 5, 1, 4, 2],
      "Prahaar": [0, 4, 5, 0, 5],
      "Kantara": [5, 4, 0, 4, 1],
      "Chhaava": [0, 5, 0, 0, 4],
      "Saiyara": [0, 0, 4, 0, 5]}

ratings = pd.DataFrame(d1, index=["U1", "U2", "U3", "U4", "U5"])
print("User-Item Rating Matrix:")
print(ratings)
```

# 1. User-Based Collaborative Filtering

```python


sim_user = cosine_similarity(ratings.values)
sim_user_df = pd.DataFrame(sim_user, index=ratings.index, columns=ratings.index)

user = "U1"
scores = {}

for item in ratings.columns:
    if ratings.loc[user, item] == 0:
        mask = ratings[item] > 0
        weights = sim_user_df.loc[user, mask]
        
        if weights.sum() > 0:
            scores[item] = (weights * ratings.loc[mask, item]).sum() / weights.sum()

print(f"\nUser-based Recommendations for {user}:")
for item, score in sorted(scores.items(), key=lambda x:x[1], reverse=True)[:2]:
    print(f"{item}: predicted rating = {score:.2f}")
```

# 2. Item-Based Collaborative Filtering


```python
sim_item = cosine_similarity(ratings.T.values)
sim_item_df = pd.DataFrame(sim_item, index=ratings.columns, columns=ratings.columns)

user = "U1"
scores = {}

for item in ratings.columns:
    if ratings.loc[user, item] == 0:
        rated = ratings.loc[user] > 0
        # Fixed the IndentationError present in the original screenshot
        weights = sim_item_df.loc[item, rated]
        
        if weights.sum() > 0:
            scores[item] = (weights * ratings.loc[user, rated]).sum() / weights.sum()

print(f"\nItem-based Recommendations for {user}:")
for item, score in sorted(scores.items(), key=lambda x:x[1], reverse=True)[:2]:
    print(f"{item}: predicted rating = {score:.2f}")
```
